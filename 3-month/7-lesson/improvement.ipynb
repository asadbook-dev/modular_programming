# Improved

```python
df=pd.read_csv("loan_data.csv")

df = df.dropna(subset=['loan_status'])

x=df.drop('loan_status', axis=1)
y=df['loan_status']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Data leakage

```python

import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

def fillna_train(df):   
        fill_values = {}
        df=df.copy()

        for col in df.columns:
            if df[col].dtype == 'object':
                fill_values[col] = df[col].mode()[0]
                df[col].fillna(fill_values[col], inplace=True)
            else:
                fill_values[col] = df[col].mean()
                df[col].fillna(fill_values[col], inplace=True)

        return df, fill_values

def fillna_test(df, fill_values):
        df=df.copy()

        for col, value in fill_values.items():
            df[col].fillna(value, inplace=True)

        return df

def encode_train(df, threshold=0):
        df = df.copy()
        encoders = {}
        onehot_cols = []

        for col in df.columns:
            if df[col].dtype == 'object':
                if df[col].nunique() <= threshold:
                    onehot_cols.append(col)
                    dummies = pd.get_dummies(df[col], prefix=col, dtype=int)
                    df = pd.concat([df.drop(columns=col), dummies], axis=1)
                else:
                    le = LabelEncoder()
                    df[col] = le.fit_transform(df[col])
                    encoders[col] = le

        return df, encoders, onehot_cols

def encode_test(df, encoders, onehot_cols, train_columns):
        df = df.copy()

        # Label encoding
        for col, le in encoders.items():
            df[col] = df[col].astype(str)
            df[col] = df[col].apply(lambda x: le.transform([x])[0] if x in le.classes_ else -1)

        # One-hot ONLY for known columns
        for col in onehot_cols:
            dummies = pd.get_dummies(df[col], prefix=col, dtype=int)
            df = pd.concat([df.drop(columns=col), dummies], axis=1)

        # Align columns
        df = df.reindex(columns=train_columns, fill_value=0)

        return df

def scale_train(df):
        df = df.copy()
        scalers = {}

        for col in df.columns:
            if df[col].dtype != 'object':
                scaler = StandardScaler()
                df[col] = scaler.fit_transform(df[[col]])
                scalers[col] = scaler

        return df, scalers

def scale_test(df, scalers):
        df = df.copy()

        for col in df.columns:
            if col in scalers:
                df[col] = scalers[col].transform(df[[col]])

        return df
    
```

```python
x_filled, fill_values = fillna_train(x_train)
x_test_filled = fillna_test(x_test, fill_values)

x_encoded, encoders, onehot_cols = encode_train(x_filled, threshold=10)
x_test_encoded = encode_test(x_test_filled, encoders, onehot_cols, x_encoded.columns)

x_scaled, scalers = scale_train(x_encoded)
x_test_scaled = scale_test(x_test_encoded, scalers)
```

# Presicion and recall


```python
# 1 => Qarz beriladi
# 0 => Qarz berilmaydi

# Qarz berilishi kerak bolmagan mijozlarga qarz berib yuborish katta risk hisoblanadi. Ya'ni 0 ni 1 predict qiladi => False Positive (FP): buni kamaytirishimiz kerak. "Precision muhim!"
```

```python
log_reg_i = LogisticRegression()
d_tree_i = DecisionTreeClassifier()
r_forest_i = RandomForestClassifier()
svm_i = SVC()

log_reg_i.fit(x_scaled, y_train)
d_tree_i.fit(x_scaled, y_train)
r_forest_i.fit(x_scaled, y_train)
svm_i.fit(x_scaled, y_train)

y_pred_log_reg_i = log_reg_i.predict(x_test_scaled)
y_pred_d_tree_i = d_tree_i.predict(x_test_scaled)
y_pred_r_forest_i = r_forest_i.predict(x_test_scaled)
y_pred_svm_i = svm_i.predict(x_test_scaled)

accuracy_log_reg_i = log_reg_i.score(x_test_scaled, y_test)
accuracy_d_tree_i = d_tree_i.score(x_test_scaled, y_test)
accuracy_r_forest_i = r_forest_i.score(x_test_scaled, y_test)
accuracy_svm_i = svm_i.score(x_test_scaled, y_test)
```

```python
from sklearn.metrics import classification_report, precision_score, recall_score

# Logistic Regression Classification Report

report_log_reg=classification_report(y_test,y_pred_log_reg_i)
print(report_log_reg)

probs = log_reg_i.predict_proba(x_test_scaled)[:, 1]   # faqat class=1 uchun ehtimollik
print("Probabilities:", probs[0])
print("Predictions:", y_pred_log_reg_i[0])

low_thr = 0.9
y_pred_low = (probs >= low_thr).astype(int)   # endi 1D bo'ladi
print("Predictions with Low Threshold:", y_pred_low[0])

print("\n=== High THRESHOLD (0.8) ===")
print("Predictions:", y_pred_low)
print("Precision:", precision_score(y_test, y_pred_low))
print("Recall:", recall_score(y_test, y_pred_low))
```

```python
print("Logistic Regression Accuracy:", accuracy_log_reg)
print("Decision Tree Accuracy:", accuracy_d_tree)
print("Random Forest Accuracy:", accuracy_r_forest)
print("SVM Accuracy:", accuracy_svm)
```
